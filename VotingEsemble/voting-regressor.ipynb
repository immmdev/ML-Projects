<h2>Voting Regressor</h2>

<p>
A <b>Voting Regressor</b> is an ensemble learning technique that combines the predictions of multiple regression models and returns the average of their predicted values.
</p>

<h3>Working</h3>
<ol>
    <li>Train multiple regression models on the same dataset.</li>
    <li>Each model predicts a numerical value.</li>
    <li>Calculate the average of all predictions.</li>
    <li>Return the average as the final prediction.</li>
</ol>

<h3>Example</h3>

<p><b>Predictions from different models:</b></p>

<ul>
    <li>Linear Regression → 10</li>
    <li>Decision Tree Regressor → 12</li>
    <li>Random Forest Regressor → 11</li>
</ul>

<p>
<b>Final Prediction:</b>
(10 + 12 + 11) / 3 = 11
</p>

<h3>Advantages</h3>
<ul>
    <li>Improves prediction accuracy.</li>
    <li>Reduces variance and overfitting.</li>
    <li>Produces more stable predictions than a single model.</li>
</ul>

<h3>Definition</h3>
<p>
A <b>Voting Regressor</b> is an ensemble method that combines the outputs of multiple regression models and predicts the final value by averaging their predictions.
</p>

```python
from sklearn.datasets import load_diabetes
from sklearn.svm import SVR
from sklearn.tree import DecisionTreeRegressor
from sklearn.linear_model import LinearRegression
import numpy as np
from sklearn.model_selection import cross_val_score

X,y=load_diabetes(return_X_y=True)
```

```python
svr=SVR()
dt=DecisionTreeRegressor()
lr=LinearRegression()

estimators=[('svr',svr),('dt',dt),('lr',lr)]
```

```python
for estimator in estimators:
    x=cross_val_score(estimator[1],X,y,cv=10,scoring='r2')
    print(estimator[0],np.round(np.mean(x),2))
```

```python
from sklearn.ensemble import VotingRegressor
vr=VotingRegressor(estimators=estimators)
x=cross_val_score(vr,X,y,cv=10,scoring='r2')
print(np.round(np.mean(x),2))
```

```python
# weighted voting
for i in range(1,4):
    for j in range(1,4):
        for k in range(1,4):
            vr=VotingRegressor(estimators=estimators,weights=[i,j,k])
            x=cross_val_score(vr,X,y,cv=10,scoring='r2')
            print("i={} j={} k={}".format(i,j,k),np.round(np.mean(x),2))
```
