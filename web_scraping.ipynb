```python
import pandas as pd
from bs4 import BeautifulSoup
import requests
```

```python
# mimic like browser as robot.txt not allow access to bots
headers = {
    "User-Agent": (
        "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
        "AppleWebKit/537.36 (KHTML, like Gecko) "
        "Chrome/137.0.0.0 Safari/537.36"
    ),
    
    "Accept": (
        "text/html,application/xhtml+xml,"
        "application/xml;q=0.9,image/avif,image/webp,*/*;q=0.8"
    ),
    
    "Accept-Language": "en-US,en;q=0.9",
    
    "Accept-Encoding": "gzip, deflate, br",
    
    "Connection": "keep-alive",
    
    "Upgrade-Insecure-Requests": "1",
    
    "Referer": "https://www.google.com/",
    
    "DNT": "1"
}
webpage=requests.get("https://www.ambitionbox.com/list-of-companies?page=1",headers=headers).text
```

```python
soup=BeautifulSoup(webpage,'lxml')
company_name=soup.find_all('h2')
for i in company_name:
    print(i.text.strip())
```

```python
soup=BeautifulSoup(webpage,'lxml')
ratings=soup.find_all('div',class_='rating_text')
for i in ratings:
    print(i.text.strip())
```

```python
soup=BeautifulSoup(webpage,'lxml')
h2=soup.find_all('span',class_='companyCardWrapper__ActionCount')
for i in h2:
    print(i.text.strip())
```

```python
soup=BeautifulSoup(webpage,'lxml')
h2=soup.find_all('div',class_='col-left')
company_name=[]
reviews=[]
ratings=[]
c_type=[]
hq=[]
# fetching big container then iterating throgh it in order to get data of all cards in a page
for i in h2:
    names=i.find_all('h2')
    for name in names:
        company_name.append(name.text.strip())
        
    all_reviews=i.find_all('span',class_='companyCardWrapper__companyRatingCount')
    for review in all_reviews:
        reviews.append(review.text.strip())
    
    all_ratings=i.find_all('div',class_='rating_text')
    for rating in all_ratings:
        ratings.append(rating.text.strip())
    
    all_c_type=i.find_all('span',class_='companyCardWrapper__interLinking')
    for c in all_c_type:
        c_type.append(c.text.strip().split('|')[0].strip())
        hq.append(c.text.strip().split('|')[1].strip().split(" ")[0])
    
# to create data frame we pass a dictionary
d={'Name':company_name,'Reviews':reviews,"Ratings":ratings,"Company_type":c_type,"HQ":hq}
```

```python
# data frame for page 1 created
df=pd.DataFrame(d)
df
```

```python
# now iterate 49 pages to fetch the complete data of companies from diff pages 
c_names=[]
c_reviews=[]
c_ratings=[]
c_types=[]
c_hqs=[]
for j in range(1,50):
    # loading diff web page in each itr
    webpage=requests.get(f"https://www.ambitionbox.com/list-of-companies?page={j}",headers=headers).text
    soup=BeautifulSoup(webpage,'lxml')
    # loading the container from each page
    conatiner=soup.find_all('div',class_='col-left')
    for i in conatiner:
        names=i.find_all('h2')
        for name in names:
            c_names.append(name.text.strip())
            
        all_c_reviews=i.find_all('span',class_='companyCardWrapper__companyRatingCount')
        for review in all_c_reviews:
            c_reviews.append(review.text.strip())
        
        all_c_ratings=i.find_all('div',class_='rating_text')
        for rating in all_c_ratings:
            c_ratings.append(rating.text.strip())
        
        all_c_types=i.find_all('span',class_='companyCardWrapper__interLinking')
        for c in all_c_types:
            c_types.append(c.text.strip().split('|')[0].strip())
            
        all_c_hqs=i.find_all('span',class_='companyCardWrapper__interLinking')
        for h in all_c_hqs:
            if(len(h.text.strip().split('|'))>1):
                c_hqs.append(h.text.strip().split('|')[1].strip().split(" ")[0])
            else:
                c_hqs.append('NaN')
```

```python
di={'Name':c_names,'Reviews':c_reviews,"Ratings":c_ratings,"Company_type":c_types,"HQ":c_hqs}
print(len(c_names),len(c_reviews),len(c_ratings),len(c_types),len(c_hqs))
```

```python
# data frame for pages 1-49 created
df=pd.DataFrame(di)
df.to_csv("company_details.csv")
```
