```python
import seaborn as sns
import pandas as pd
import matplotlib as plt
```

```python
df=pd.read_csv('Titanic-Dataset.csv')
# now we will study all the columns one by one
```

## 1. Categorical data
<p> If you are dealing with the categorical data then you have generally two options to analyse the columns left to right, in univariate analysis.</p>

```python
df['Survived'].value_counts().plot(kind='bar')
```

```python
sns.countplot(x=df['Pclass'],data=df)
```

```python
sns.countplot(x=df['Sex'],data=df)
```

```python
sns.countplot(x=df['Embarked'],data=df)
```

```python
df['Embarked'].value_counts().plot(kind='pie')
```

## 2. Numerical data

```python
df['Age'].plot(kind='hist')
```

```python

df.plot.hist(column='Age',bins=50)
```

```python
sns.displot(x=df["Age"])
```

```python
df["Age"].plot(kind='box')
```

```python
sns.boxplot(x=df["Age"])
```
