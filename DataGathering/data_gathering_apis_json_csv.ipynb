```python
import requests
import pandas as pd
```

```python
df=pd.DataFrame()
```

```python
headers = {
    "Authorization": "Bearer <your_token>",
    "accept": "application/json"
}

# creating data of 2000 rows by fetching top 100 pages with 20 rows each containing name all
for i in range(1,101):
    url = f"https://api.themoviedb.org/3/search/movie?query=all&page={i}"
    res = requests.get(url, headers=headers)
    data=res.json()
    # converting json into a dataframe
    df_temp=pd.DataFrame(data['results'])[['title','original_language','original_title','overview','popularity','poster_path','release_date','vote_average','vote_count']]
    # concatening empty df with new one
    df=pd.concat([df,df_temp],ignore_index=True)
```

```python
# coverting to csv file 
df.to_csv("movies.csv")
```
