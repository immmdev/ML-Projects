# Encoding

Encoding is the process of converting categorical data into numerical data so that machine learning models can understand and process it effectively.

There are two types of categorical data:

1. **Ordinal Data**
   Data that has a hierarchical or meaningful order.
   For this type of data, we use **Ordinal Encoding**.

   Example:

   * Low < Medium < High
   * High School < Bachelor < Master < PhD

2. **Nominal Data**
   Data that has no meaningful order or hierarchy.
   For this type of data, we use **One Hot Encoding**.

   Example:

   * Red, Blue, Green
   * Delhi, Mumbai, Chennai

**Note:**
For the target/output column, especially in classification problems, we generally use **Label Encoding**.


```python
import pandas as pd
```

```python
data = {
    "Name": ["Alice", "Bob", "Charlie", "David", "Eva"],
    "Education": ["High School", "Bachelor", "Master", "PhD", "Bachelor"],
    "City": ["Delhi", "Mumbai", "Delhi", "Bangalore", "Chennai"],
    "Gender": ["Female", "Male", "Male", "Male", "Female"],
    "Experience_Level": ["Beginner", "Intermediate", "Advanced", "Expert", "Intermediate"]
}
df=pd.DataFrame(data)
```

```python
from sklearn.preprocessing import OrdinalEncoder
oe=OrdinalEncoder(categories=[['High School','Bachelor','Master','PhD']])
# ordinal encoding
df['Education']=oe.fit_transform(df[['Education']])
```

```python
# label encoding (espicially for output column)
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
df['Experience_Level']=le.fit_transform(df['Experience_Level'])
```

```python
df['Education']=df["Education"].astype(int)
```

```python
# one hot encoding for nominal data
from sklearn.preprocessing import OneHotEncoder
ohe=OneHotEncoder(drop='first',sparse_output=False)
encoded=ohe.fit_transform(df[['Gender']])
```

```python
encoded_df=pd.DataFrame(encoded,columns=ohe.get_feature_names_out(['Gender']))
```
