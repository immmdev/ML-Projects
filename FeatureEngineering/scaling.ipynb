# Feature Scaling in Machine Learning

## What is Feature Scaling?

Feature Scaling is the process of transforming numerical features into a similar range so that machine learning models can learn efficiently.

Some algorithms are highly sensitive to feature magnitude, such as:

* KNN
* K-Means
* Logistic Regression
* Linear Regression
* SVM
* Neural Networks

Example:

| Age | Salary |
| --- | ------ |
| 22  | 20000  |
| 35  | 150000 |

Here, Salary has much larger values than Age.
Without scaling, the model may give more importance to Salary.

---

# Types of Scaling

## 1. Standardization (StandardScaler)

### Definition

Standardization transforms data so that:

* Mean becomes `0`
* Standard deviation becomes `1`

### Formula x'<sub>i</sub>=(x<sub>i</sub>-μ)/σ

Where:

* `x` = original value
* `μ` = mean
* `σ` = standard deviation

---

## Characteristics

* Output range is not fixed
* Can contain negative values
* Works well when data follows normal distribution

---

## When to Use?

Use StandardScaler when:

* data is approximately normally distributed
* algorithms depend on distance or gradient

Commonly used in:

* Logistic Regression
* Linear Regression
* PCA
* SVM
* Neural Networks

---

## Advantages

* Centers the data
* Improves gradient descent convergence
* Handles moderate outliers better than MinMaxScaler

---

## Disadvantages

* Sensitive to extreme outliers

---

## Syntax

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

scaled_data = scaler.fit_transform(df)
```

---

# 2. Min-Max Scaling (Normalization)

## Definition

Min-Max Scaling transforms data into a fixed range, usually between `0` and `1`.

### Formula

x' = \frac{x-x_{min}}{x_{max}-x_{min}}

---

## Characteristics

* Values lie between `0` and `1`
* Preserves original distribution shape
* Very sensitive to outliers

---

## When to Use?

Use MinMaxScaler when:

* data does not follow normal distribution
* bounded range is needed

Commonly used in:

* Image Processing
* Neural Networks
* Deep Learning

---

## Advantages

* Keeps values in fixed range
* Easy to interpret
* Useful for gradient-based algorithms

---

## Disadvantages

* Highly affected by outliers

Example:
If one salary is `10,00,000`, all other values become compressed.

---

## Syntax

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

scaled_data = scaler.fit_transform(df)
```

---

# 3. Robust Scaling (RobustScaler)

## Definition

RobustScaler uses:

* Median
* Interquartile Range (IQR)

instead of mean and standard deviation.

### Formula

x' = \frac{x-\text{Median}}{IQR}

Where:

IQR = Q3 - Q1

* `Q1` = 25th percentile
* `Q3` = 75th percentile

---

## Characteristics

* Less sensitive to outliers
* Uses median instead of mean

---

## When to Use?

Use RobustScaler when:

* dataset contains many outliers
* extreme values affect scaling

---

## Advantages

* Handles outliers effectively
* Stable scaling

---

## Disadvantages

* Does not guarantee fixed range
* Less useful when data is normally distributed

---

## Syntax

```python
from sklearn.preprocessing import RobustScaler

scaler = RobustScaler()

scaled_data = scaler.fit_transform(df)
```

---

# Comparison Table

| Scaler         | Uses Mean | Uses Std | Handles Outliers | Output Range   |
| -------------- | --------- | -------- | ---------------- | -------------- |
| StandardScaler | Yes       | Yes      | Moderate         | No fixed range |
| MinMaxScaler   | No        | No       | Poor             | 0 to 1         |
| RobustScaler   | No        | No       | Excellent        | No fixed range |

---

# Important Notes

## StandardScaler

* Best for normally distributed data

## MinMaxScaler

* Best when fixed range is required

## RobustScaler

* Best when dataset contains outliers

---

# Quick Interview Difference

## StandardScaler vs MinMaxScaler

| StandardScaler             | MinMaxScaler                 |
| -------------------------- | ---------------------------- |
| Mean = 0, Std = 1          | Values between 0 and 1       |
| Can have negative values   | No negative values usually   |
| Less sensitive to outliers | Highly sensitive to outliers |

---

# Real-Life Intuition

Suppose heights are:

```python
[150, 160, 170, 180]
```

## MinMaxScaler

Converts roughly into:

```python
[0, 0.33, 0.66, 1]
```

## StandardScaler

Converts around mean:

```python
[-1.34, -0.44, 0.44, 1.34]
```

## RobustScaler

Scales using median and IQR, reducing outlier impact.

---

# Conclusion

Feature scaling is important because many ML algorithms depend on feature magnitude.

Choose scaler based on:

* data distribution
* presence of outliers
* algorithm requirements


````python
import pandas as pd

data = {
    "Age": [18, 22, 25, 30, 35, 40, 45],
    "Salary": [15000, 25000, 40000, 60000, 85000, 120000, 180000],
    "Height_cm": [155, 160, 165, 170, 175, 180, 185],
    "Weight_kg": [50, 55, 60, 68, 75, 82, 90]
}

df = pd.DataFrame(data)

print(df)
````

````python
from sklearn.preprocessing import StandardScaler,MinMaxScaler
ss=StandardScaler()
mm=MinMaxScaler()
df['Salary'].dtype
````

````python
for column in df:
    if df[column].dtype=='int64':
        df[column]=ss.fit_transform(df[[column]])
````

````python
for column in df:
    if df[column].dtype=='int64':
        df[column]=mm.fit_transform(df[[column]])
````
