<h1>AdaBoost (Adaptive Boosting)</h1>

<p>
AdaBoost is an ensemble boosting algorithm that combines multiple weak learners
to create one strong learner.
</p>

<hr>

<h2>Core Idea</h2>

<ul>
  <li>Train weak learners sequentially</li>
  <li>Each new learner focuses more on previous mistakes</li>
  <li>Combine all learners using weighted voting</li>
</ul>

<hr>

<h2>Weak Learner</h2>

<p>
Usually a Decision Stump:
</p>

<pre>
Decision Tree with max_depth = 1
</pre>

<hr>

<h2>Why Called Adaptive?</h2>

<p>
Because sample weights are updated after every iteration.
Wrongly classified samples receive higher importance.
</p>

<hr>

<h2>Working of AdaBoost</h2>

<h3>Step 1: Initialize Weights</h3>

<pre>
w_i = 1 / n
</pre>

<p>
All samples initially get equal weights.
</p>

<hr>

<h3>Step 2: Train Weak Learner</h3>

<p>
Train a small decision tree (stump) on the weighted dataset.
</p>

<hr>

<h3>Step 3: Calculate Error</h3>

<pre>
Error = Sum of wrongly classified weights
</pre>

<hr>

<h3>Step 4: Calculate Alpha (Model Importance)</h3>

<pre>
alpha = 0.5 * ln((1-error)/error)
</pre>

<ul>
  <li>Lower error → Higher alpha</li>
  <li>Better model → More voting power</li>
</ul>

<hr>

<h3>Step 5: Update Weights</h3>

<pre>
w_i = w_i * e^(-alpha * y_i * h(x_i))
</pre>

<ul>
  <li>Wrong predictions → weight increases</li>
  <li>Correct predictions → weight decreases</li>
</ul>

<hr>

<h3>Step 6: Normalize Weights</h3>

<p>
Convert weights into probabilities so total sum becomes 1.
</p>

<hr>

<h3>Step 7: Repeat</h3>

<p>
Train next learner using updated weights.
</p>

<hr>

<h2>Final Prediction</h2>

<pre>
F(x) = Σ(alpha_m * h_m(x))
</pre>

<ul>
  <li>Classification → Weighted voting</li>
  <li>Regression → Weighted averaging</li>
</ul>

<hr>

<h2>Why Exponential Function is Used?</h2>

<ul>
  <li>Smooth scaling of weights</li>
  <li>Strong emphasis on difficult samples</li>
  <li>Multiplicative updates instead of additive updates</li>
  <li>Works well for both small and large alpha values</li>
</ul>

<hr>

<h2>Exponential Loss</h2>

<pre>
L = e^(-yF(x))
</pre>

<p>
AdaBoost minimizes exponential loss internally.
</p>

<hr>

<h2>Boosting Concept</h2>

<p>
Boosting means:
</p>

<ul>
  <li>Models are trained sequentially</li>
  <li>Each learner corrects previous mistakes</li>
  <li>Weak learners combine into a strong learner</li>
</ul>

<h2>Advantages</h2>

<ul>
  <li>High accuracy</li>
  <li>Improves weak learners</li>
  <li>Works well on tabular data</li>
  <li>Reduces bias</li>
</ul>

<hr>

<h2>Disadvantages</h2>

<ul>
  <li>Sensitive to outliers</li>
  <li>Sequential training is slower</li>
  <li>Can overfocus on noisy samples</li>
</ul>

<ul>

```python
from sklearn.ensemble import AdaBoostClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score
from sklearn.model_selection import GridSearchCV
```

```python
import pandas as pd
df=pd.read_csv('heart.csv')
X=df.iloc[0:500,0:-1]
y=df.iloc[0:500,-1]
```

```python
base_model=DecisionTreeClassifier(max_depth=1) # decision stump
model=AdaBoostClassifier(estimator=base_model) # adaboost instance
# fine tuning using gridcbsearch
param_grid={
    'learning_rate':[.1,.2,.3,.4,.5],
    'n_estimators':[100,200,400,500]
}

gc=GridSearchCV(estimator=model,param_grid=param_grid,cv=5,verbose=2,n_jobs=-1)
# training
gc.fit(X,y)
```

```python
best_model=gc.best_estimator_
pred=best_model.predict(df.iloc[100:1000,:-1])
print("accuracy is {} %".format(accuracy_score(pred,df.iloc[100:1000,-1])*100))
```
